# Fruit Store Management System

### Class - FruitStore

__init__ method                        --> Displays welcome message  

Load_dataset(option) Method			   --> Loads dataset, and save the changes made  

menu() method						   --> Display fruit and menu options  

get_condition(fruit_id) method         --> Display quality information of fruit  

check_stockinfo(fruit_id) method	   --> Display stock availability and pricing for the selected fruit  

show_fruitinfo(fruit_id) method		   --> Display description and origin of fruit  


### Python libraries used:
pandas, sys, csv

### Data Used:
fruit_store_dataset.csv

fruit_id  

fruit_name  

condition - A: Top Quality  

			B: Moderate Quality  
   
			C: Poor Quality  
   
in_stock_availability - Stock level (High, Medium, Low)  

buy_price  

sell price  

description  

origin

```python
import pandas as pd
import sys
import csv

class FruitStore():
    def __init__(self):
        """
        Initialize the fruit store: display welcome message and load dataset.
        """
        print("Welcome to the Toronto Fruit Store")
        print("\nThe only place to find DeLiCiOuS and hEaLtHy Food from around the world\n")
        self.Load_dataset(1)  # Load dataset and start menu

    def Load_dataset(self, option):
        """
        Load the dataset from CSV file or exit/save changes.
        option: 1 -> load dataset
                2 -> save dataset and exit
        """
        filename = 'fruit_store_dataset.csv'   # Make sure your CSV file is named like this
        
        if option == 1:
            # Load CSV into DataFrame
            self.df = pd.read_csv(filename)
            # Create a dictionary mapping fruit_id to fruit_name for display
            self.fruit_dict = dict(zip(self.df["fruit_id"], self.df["fruit_name"]))
            # Start the menu
            self.menu()
        elif option == 2:
            # Save the DataFrame to CSV and exit
            self.df.to_csv(filename, index=False)
            print("\nThank you for Shopping with us. Goodbye!")
            sys.exit()
        else:
            pass

    def menu(self):
        """
        Display menu of fruits and options for the user.
        """
        print("\n--- Fruit Available in Store ---\n")
        print(self.fruit_dict)
        print("\nPress E to Exit\n")
        
        fruit_input = input("Enter your choice (Fruit Id): ").strip()
        
        # Exit condition
        if fruit_input.lower() == 'e':
            self.Load_dataset(2)
        
        # Try to convert input to integer
        try:
            fruit_id = int(fruit_input)
            
            # Check if fruit_id is valid (1-100 assumed)
            if fruit_id in range(1, 101):
                # Display options menu
                options_menu = {
                    1: 'Get Fruit Details',
                    2: 'Check Stock Details',
                    3: 'Show Fruit Description',
                    4: 'Quit'
                }
                print(options_menu)
                
                choice = int(input("Enter your choice (1 2 3 4): "))
                
                # Call corresponding method
                if choice == 1:
                    self.get_condition(fruit_id)
                elif choice == 2:
                    self.check_stockinfo(fruit_id)
                elif choice == 3:
                    self.show_fruitinfo(fruit_id)
                elif choice == 4:
                    self.Load_dataset(2)
                else:
                    print("Enter a valid choice.")
                    self.menu()
            else:
                print("Enter a valid fruit ID.")
                self.menu()
                
        except ValueError:
            # Handle non-integer input
            print("Enter a valid integer or E to exit.")
            self.menu()

    def get_condition(self, fruit_id):
        """
        Display the quality condition of the selected fruit.
        """
        # Filter DataFrame for the selected fruit
        result = self.df[self.df["fruit_id"] == fruit_id]
        
        fruit_name = result.iloc[0]["fruit_name"]
        condition = result.iloc[0]["condition"]
        
        # Display fruit condition
        print(f"\nFruit Name: {fruit_name}")
        if condition == 'A':
            print("Quality: Good")
        elif condition == 'B':
            print("Quality: Moderate")
        elif condition == 'C':
            print("Quality: Poor")
        
        # Option for more info
        val = input("More Information? Press 1: ").strip()
        if val == '1':
            self.show_fruitinfo(fruit_id)
        
        self.menu()

    def check_stockinfo(self, fruit_id):
        """
        Display stock and pricing information for the selected fruit.
        """
        result = self.df[self.df["fruit_id"] == fruit_id]
        
        fruit_name = result.iloc[0]["fruit_name"]
        condition = result.iloc[0]["condition"]
        in_stock = result.iloc[0]["in_stock_availability"]
        buy_price = result.iloc[0]["buy_price"]
        sell_price = result.iloc[0]["sell_price"]
        
        # Display basic info
        print(f"\nFruit Name: {fruit_name}")
        print(f"Buy Price (CAD/kg): {buy_price}")
        print(f"Original Sell Price (CAD/kg): {sell_price}")
        print(f"Stock Availability: {in_stock}")
        
        # Adjust sell price based on condition and stock
        if condition == 'A':
            print("Condition - Top Quality")
            if in_stock == 'High':
                marked_price = sell_price
            elif in_stock == 'Medium':
                marked_price = sell_price * 1.25
            elif in_stock == 'Low':
                marked_price = sell_price * 1.5
        elif condition == 'B':
            print("Condition - Moderate Quality")
            if in_stock == 'High':
                marked_price = sell_price
            elif in_stock == 'Medium':
                marked_price = sell_price * 1.1
            elif in_stock == 'Low':
                marked_price = sell_price * 1.2
        elif condition == 'C':
            print("Condition - Poor Quality")
            if in_stock == 'High':
                marked_price = sell_price
            elif in_stock == 'Medium':
                marked_price = sell_price * 1.1
            elif in_stock == 'Low':
                marked_price = sell_price * 1.2
        
        print(f"Marked Sell Price (CAD/kg): {marked_price}")
        self.menu()

    def show_fruitinfo(self, fruit_id):
        """
        Display description and origin of the selected fruit.
        """
        result = self.df[self.df["fruit_id"] == fruit_id]
        
        fruit_name = result.iloc[0]["fruit_name"]
        description = result.iloc[0]["description"]
        origin = result.iloc[0]["origin"]
        
        print(f"\nFruit Name: {fruit_name}")
        print(f"Description: {description}")
        print(f"Origin: {origin}")
        print("-----*******-------*******")
        
        self.menu()

# To start the store
# fstore = FruitStore()
```
