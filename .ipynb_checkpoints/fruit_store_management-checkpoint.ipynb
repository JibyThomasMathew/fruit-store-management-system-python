# Fruit Store Management System

Create a class FruitStore that should:

## Functions (Methods):

    show_menu() → Displays fruit options with IDs
    get_choice() → Asks user to enter fruit ID
    get_color(fruit_id) → Returns fruit color
    get_price(fruit_id) → Returns fruit price
    check_stock(fruit_id) → Returns quantity available
    show_summary(fruit_id) → Prints all information clearly

## Data Dictionaries (Inside the class):
colors = {1:"Yellow", 2:"Red", 3:"Green", 4:"Orange"}
prices = {1:300, 2:200, 3:500, 4:250}
stock  = {1:20, 2:0,   3:10, 4:50}

## Task:

Build a console program that:
    Shows the menu
    Accepts a fruit choice from the user
    Shows color, price, and availability
    Prints “Available” or “Out of Stock”

Bonus Challenge:
Add a method to continue shopping until the user chooses to exit.

```python
import pandas as pd
import sys
import csv

class FruitStore():
    def __init__(self):
        print("Welcome to the Toronto Fruit Store")
        print("\n\n\nThe only place to find DeLiCiOuS and hEaLtHy Food from around the world")
        self.Load_dataset(1) 

    def Load_dataset(self, i):
        filename = 'fruit_store_dataset.csv'   # ensure file is renamed to this
    
        if i == 1:
            self.df = pd.read_csv(filename)
            self.fruit_dict = dict(zip(self.df["fruit_id"], self.df["fruit_name"]))
            self.menu() 
        elif i == 2:
            print("\n\n\n\n\n\n\n\n\nThank you for banking with us. \nBye \n\n\n\n\n\n\n")
            self.df.to_csv(filename, index=False)
            print(f"All changes saved to {filename} successfully!")
            sys.exit()
        else:
            pass
    
    def menu(self):

        print("\n\n\n\nFruit Available in Store")

        print("\n\n\n\n\nChoose the fruit\n\n\n\n\n\n")
        print(self.fruit_dict)

        self.fruit_id = int(input("Enter your choice, Fruit Id\n\n\n\n"))
        if self.fruit_id in range(1,101):
        
            menu = {1:'Get Fruit Details',
                    2:'Check Stock Details',
                    3:'Show Fruit Description',
                    4:'Quit'}
            print(menu)
            val_temp1 = int(input("\n\n\n\n\nEnter your choice : 1   2   3   4  :"))

            if val_temp1 == 1:
                self.get_condition(self.fruit_id)
            elif val_temp1 == 2:
                self.check_stockinfo(self.fruit_id)
            elif val_temp1 == 3:
                self.show_fruitinfo(self.fruit_id)
            elif val_temp1 == 4:
                self.Load_dataset(2)
            else:
                print(" Enter the correct input : ")
                self.menu()

        else:
            print(" Enter the correct input : ")
            self.menu()
         
            
        
    def get_condition(self, fruit_id):
        result = self.df[self.df["fruit_id"] == self.fruit_id]

        # Extract values
        fruit_name = result.iloc[0]["fruit_name"]
        condition = result.iloc[0]["condition"]

        # Display output
        print(f"Fruit Name: {fruit_name}")
        if condition == 'A':
            print("Quality:Good")
        elif condition == 'B':
            print("Quality: Moderate")
        elif condition == 'C':
            print("Quality:Poor")

        val_temp3 = int(input("\n\n\n\nMore Information, Press 1 :"))
        if val_temp3 == 1:
            self.show_fruitinfo(self.fruit_id)
        self.menu()       

    def check_stockinfo(self, fruit_id):
        result = self.df[self.df["fruit_id"] == self.fruit_id]
        
        fruit_id = result.iloc[0]["fruit_id"]
        fruit_name = result.iloc[0]["fruit_name"]
        condition = result.iloc[0]["condition"]
        in_stock_availability = result.iloc[0]["in_stock_availability"]
        buy_price = result.iloc[0]["buy_price"]
        sell_price = result.iloc[0]["sell_price"]

        print(f"Fruit Name: {fruit_name}")
        print(f"Buy Price (CAD/kg): {buy_price}")
        print(f"Original Sell Price (CAD/kg): {sell_price}")
        print(f"Stock Availability: {in_stock_availability}")
        
        if condition == 'A':
            print("Condition - Top Quality")
            if in_stock_availability == 'High':
                print(f"Marked Sell Price (CAD/kg): {sell_price}")
            elif in_stock_availability == 'Medium':
                print(f"Marked Sell Price @125% (CAD/kg): {sell_price*1.25}")
            elif in_stock_availability == 'Low':
                print(f"Marked Sell Price @150% (CAD/kg): {sell_price*1.50}")
        elif condition == 'B':
            print("Condition - Moderate Quality")
            if in_stock_availability == 'High':
                print(f"Marked Sell Price (CAD/kg): {sell_price}")
            elif in_stock_availability == 'Medium':
                print(f"Marked Sell Price @110% (CAD/kg): {sell_price*1.10}")
            elif in_stock_availability == 'Low':
                print(f"Marked Sell Price @120% (CAD/kg): {sell_price*1.20}")
        elif condition == 'C':
            print("Condition - Poor Quality")
            if in_stock_availability == 'High':
                print(f"Marked Sell Price (CAD/kg): {sell_price}")
            elif in_stock_availability == 'Medium':
                print(f"Marked Sell Price @110% (CAD/kg): {sell_price*1.10}")
            elif in_stock_availability == 'Low':
                print(f"Marked Sell Price @120% (CAD/kg): {sell_price*1.20}")

        self.menu()
          
    def show_fruitinfo(self, fruit_id):
        result = self.df[self.df["fruit_id"] == self.fruit_id]
        
        # Extract values
        fruit_name = result.iloc[0]["fruit_name"]
        description = result.iloc[0]["description"]
        origin = result.iloc[0]["origin"]
        
        # Display output
        print(f"Fruit Name: {fruit_name}")
        print(f"Description: {description}")
        print(f"Origin: {origin}")
        print("-----*******-------*******")

        self.menu()

```
